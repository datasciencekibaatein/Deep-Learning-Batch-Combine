```python
import numpy as np
```

input

```python
x = np.array([-0.3,-0.1,-0.2,-0.2])
```

# 1 hidden layers

weight for input

```python
w1 = np.array(
    [
        [0.3,0.4,0.3],
        [0.5,0.2,0.3],
        [0.2,0.6,0.2],
        [0.1,0.3,0.6]
    ]
)

w1_t = w1.T
```

```python
b1 = np.array([[0.1,0.1,0.1]])
```

# 2 hidden layer

```python
w2 = np.array(
    [
        [0.4,0.6],
        [0.3,0.7],
        [0.8,0.2]
    ]
)

w2_t = w2.T
```

```python
b2 = np.array([0.1,0.1])
```

Output layer

```python
w3 = np.array(
    [
        [0.4],
        [0.5]
    ]
)

w3_t = w3.T
```

```python
b3 = np.array([[0.1]])
```

## sigmoid function

```python
def sigmoid(x):
  return 1 / (1 + np.exp(-x))
```

```python

print(f"{sigmoid(100):.8f}")
```

# Forward propagate

```python
x
```

```python
w1_t
```

```python
o1 = w1_t @ x + b1
```

```python
o1_sig = sigmoid(o1)
```

next layer

```python
o1_sig[0]
```

```python
o2 = w2_t.dot(o1_sig[0]) + b2
```

```python
o2
```

```python
o2_sig = sigmoid(o2)

o2_sig
```

Next layer

```python
o3 = w3_t @ o2_sig + b3
```

```python
y_hat = sigmoid(o3)
```

```python
y_hat
```

```python
True if y_hat > 0.5 else False
```
