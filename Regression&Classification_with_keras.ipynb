```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

2 Features -> Dense Layer(4 neurons) -> Dense Layer(1 neuron)

```python
x = tf.constant([[2.0,3.0],
 [1.0,4.0],
  [3.0,2.0],
   [4.0,1.0]])

y = tf.constant([10.0,9.0,11.0,12.0])
```

```python
model = keras.Sequential(
    [
        layers.Dense(4, input_shape=(2,),activation='relu'),
        layers.Dense(1)
    ]
)
```

```python
model.compile(optimizer='sgd',loss='mse')
```

```python
history = model.fit(x,y,epochs = 100,verbose=1)
```

```python
y_pred = model.predict(x)
```

```python
model.layers[0].get_weights()
```

```python
model.layers[1].get_weights()
```

```python
model.trainable_variables
```

```python
from sklearn.metrics import mean_squared_error as mse
```

```python
with tf.GradientTape() as tape:

  y_pred = model(x)
  loss = tf.reduce_mean(tf.square(y_pred-y))
gradients = tape.gradient(loss,model.trainable_variables)
```

```python
from sklearn.datasets import make_classification


data = make_classification(1000,n_features=4,n_classes=2)
```

```python
X = data[0]

y = data[1]
```

```python
model = keras.Sequential(
    [
        layers.Dense(4,input_shape=(4,),activation='relu'),
        layers.Dense(4,activation='relu'),
        layers.Dense(4,activation='relu'),
        layers.Dense(1,activation='sigmoid')
    ]
)
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model.fit(X,y,epochs=100,verbose=1)
```

```python
for layer in model.layers:
  weights = layer.get_weights()
  print(weights)
```

```python
from sklearn.datasets import make_regression

data = make_regression(1000,n_features=4,n_targets=1)


X = data[0]
y = data[1]
```

```python
model = keras.Sequential(
    [
        layers.Dense(4,input_shape=(4,),activation='relu'),
        layers.Dense(4,activation='relu'),
        layers.Dense(4,activation='relu'),
        layers.Dense(1)
    ]
)
```

```python
model.compile(optimizer='sgd',loss='mse')
```

```python
model.fit(X,y,epochs=100,verbose=2)
```

```python
loss_fn = tf.keras.losses.MeanSquaredError()
optimizer = tf.keras.optimizers.SGD(learning_rate=0.1)
with tf.GradientTape() as tape:

  y_pred = model(X)
  loss = loss_fn(y,y_pred)
gradients = tape.gradient(loss,model.trainable_variables)

optimizer.apply_gradients(zip(gradients,model.trainable_variables))
```
