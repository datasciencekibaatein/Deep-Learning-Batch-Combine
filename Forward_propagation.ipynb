# Forward Propagation

```python
import numpy as np
```

## Layer 01

```python
x = np.array([6,8,4,0])
```

```python
w = np.array([
    [0.2,0.7,0.1],
    [0.3,0.2,0.5],
    [0.6,0.0,0.4],
    [0.7,0.1,0.2]

])

w1_t = w.T
```

```python
w1_t
```

```python
b = np.array([3,5,7])
```

```python
o1 = w1_t@x + b
```

```python
o1
```

## Layer 02

```python

w2 = np.array(
    [
        [0.3,0.7],
        [0.5,0.5],
        [0.1,0.9]
    ]
)

w_t2 = w2.T
```

```python
w_t2
```

```python
o1
```

```python
b2 = np.array([8,3])
```

```python
o2 = w_t2@o1 + b2
```

```python
o2
```

## Layer 03

```python
w3 = np.array([
    [0.8],
    [0.2]
])

w3_t = w3.T
```

```python
w3_t
```

```python
o2
```

```python
b = 5
```

```python
y_hat = w3_t@o2 + b
```

```python
print("Output: ",y_hat)
```
