# Dropout

Suppose a hidden layer has 100 neurons. During training, a dropout rate of 0.3 randomly disables approximately 30% of units one each training step.

This forces the network to learn useful patterns without depending too heavily on a fixed group of neurons.

Dropout is regularization techineque designed to help reduce overfitting. it does not permanently remove neurons form the network

```python
import numpy as np

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
X = np.random.randn(1200,10).astype("float32")

y = (X[:,0] + X[:,1] - X[:,2] > 0).astype("float32")
```

```python
X_train,X_val = X[:900],X[900:]
y_train,y_val = y[:900],y[900:]
```

```python
model = keras.Sequential([
    layers.Input(shape=(10,)),
    layers.Dense(128,activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(64,activation='relu'),
    layers.Dropout(0.2),
    layers.Dense(1,activation='sigmoid')
])
```

```python
model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(X_train,y_train,epochs=100,batch_size=30,validation_data=(X_val,y_val))
```
