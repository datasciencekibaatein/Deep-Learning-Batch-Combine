# Regularization

A simple model might learn meaning full relationships between attendance, study hours, and marks. An overfit model might instead assign excessively large weights to small fluctuations or noise in the traing data.


## L1 and L2 regularization

#### L1 regularization(Lasso like penalty)

Penalize the absolute values of weights, It push some weights exactly to zero,effectively reducing the number of active connection.

#### L2 regularization(Ridge like penalty)

Penalizes squared weights, encuraging them to remain small rather than allowing a few weights to become excessively large.

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers,regularizers

```

```python
X = np.random.randn(1000,10).astype(np.float32)
y = (X[:,0] + 0.8 * X[:,1]- 0.5 * X[:,2]>0).astype("float32")
```

```python
model = keras.Sequential([
    layers.Input(shape=(10,)),
    layers.Dense(
        32,
        activation="relu",
        kernel_regularizer=regularizers.l2(0.01)
    ),
    layers.Dense(
        16,
        activation="relu",
        kernel_regularizer=regularizers.l2(0.01)
    ),
    layers.Dense(
        1,
        activation="sigmoid"
    )
])
```

```python
model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.fit(X,y,epochs=100,verbose=1,batch_size=32)
```
