# Backpropagation

```python
import math

x1 = 1
x2 = 0
y = 1


#input to hidden layer

w1 = 0.5
w2 = 0.4

w3 = 0.3
w4 = 0.2

b1 = 0.1
b2 = 0.1

#hidden -> output layer

w5 = 0.7
w6 = 0.6

b3 = 0.1


lr = 0.5


def sigmoid(z):
  return 1 / (1 + math.exp(-z))



# forward propagation


z1 = x1 * w1 + x2 * w2 + b1
h1 = sigmoid(z1)


z2 = x1 * w3 + x2 * w4 + b2
h2 = sigmoid(z2)


z3 = h1 * w5 + h2 * w6 + b3

y_hat = sigmoid(z3)
```

```python
y_hat
```

```python
loss = 0.5 * (y - y_hat)**2

loss
```

```python
def sigmoid_derivative(a):
  return a * (1 - a)
```

# Backpropagation

```python
#dL / dy_hat

dL_dy_hat = y_hat - y

dL_dy_hat
```

```python
dyhat_dz3 = sigmoid_derivative(y_hat)

dL_dz3 = dL_dy_hat * dyhat_dz3

dL_dz3

```

```python
dL_dw5 = dL_dz3 * h1
dL_dw6 = dL_dz3 * h2

dL_dw5, dL_dw6
```

```python
dL_dh1 = dL_dz3 * w5
dL_dh2 = dL_dz3 * w6

dL_dh1, dL_dh2
```

```python
dh2_dz2 = sigmoid_derivative(h2)
dh1_dz1 = sigmoid_derivative(h1)
```

```python
dL_dz1 = dL_dh1 * dh1_dz1
dL_dz2 = dL_dh2 * dh2_dz2

dL_dz1, dL_dz2
```

```python
dL_dw1 = dL_dz1 * x1
dL_dw2 = dL_dz1 * x2
dL_dw3 = dL_dz2 * x1
dL_dw4 = dL_dz2 * x2

dL_dw1, dL_dw2, dL_dw3, dL_dw4
```

```python
w1 = w1 - lr * dL_dw1
w2 = w2 - lr * dL_dw2
w3 = w3 - lr * dL_dw3
w4 = w4 - lr * dL_dw4
w5 = w5 - lr * dL_dw5
w6 = w6 - lr * dL_dw6

b1 = b1 - lr * dL_dz1
b2 = b2 - lr * dL_dz2
b3 = b3 - lr * dL_dz3

w1, w2, w3, w4, w5, w6, b1, b2, b3
```
