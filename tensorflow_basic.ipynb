#### Matrix multiplication

```python
A = tf.constant(
    [
        [3,4,2],
        [4,2,4]
    ]
)

B = tf.constant(
    [
        [5,6],
        [6,9],
        [3,2]
    ]
)


print(tf.matmul(A,B))
```

Neural networks parameters are variables (weight and bias)

## Manual Neural network

```python
x = tf.constant([2.0,3.0])

w = tf.Variable([0.5,0.8])

b = tf.Variable(0.2)
```

```python
z = tf.reduce_sum(x*w) + b
```

```python
z
```

```python
output = tf.nn.relu(z)
```

```python
output
```

## Automatic Differentiation

```python
x = tf.Variable(3.0)

with tf.GradientTape() as tape:
  y = x ** 2


gradient = tape.gradient(y,x)

print(gradient)
```
